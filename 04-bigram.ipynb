```python
import math
import torch
```

```python
with open("04-input.txt", "r") as f:
    data = f.read()
chars = sorted(set(data))
vocab_size = len(chars)
ctoi = {c: i for i, c in enumerate(chars)}
itoc = {i: c for i, c in enumerate(chars)}
```

```python
train = data[:int(len(data) * 0.9)]
val = data[int(len(data) * 0.9):]
```

```python
N = torch.ones((vocab_size, vocab_size), dtype=torch.int32)
for ch1, ch2 in zip(train, train[1:]):
    N[ctoi[ch1], ctoi[ch2]] += 1
P = N.float() / N.sum(1, keepdim=True)
```

```python
output = ['\n']
max_output_length = 1000
while len(output) < max_output_length:
    i = ctoi[output[-1]]
    j = torch.multinomial(P[i], num_samples=1, replacement=True).item()
    output.append(itoc[j])
```

```python
''.join(output)
```

```python
def negative_log_likelihood(P, data):
    val = 0
    count = 0
    for ch1, ch2 in zip(data, data[1:]):
        val -= math.log(P[ctoi[ch1], ctoi[ch2]])
        count += 1
    val /= count
    return val

negative_log_likelihood(P, train), negative_log_likelihood(P, val)
```
