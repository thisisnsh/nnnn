```python
import math
import torch
```

```python
# input & output initialization
with open(file="inputs-shakespeare.txt", mode="r", encoding="utf-8") as f:
    data = f.read()

chars = sorted(set(data))
vocab_size = len(chars)
ctoi = {char: i for i, char in enumerate(iterable=chars)}
itoc = dict(enumerate(iterable=chars))

train = data[:int(len(data) * 0.9)]
val = data[int(len(data) * 0.9):]
```

```python
# build count model
N = torch.ones(size=(vocab_size, vocab_size), dtype=torch.int32)
for ch1, ch2 in zip(train, train[1:]):
    N[ctoi[ch1], ctoi[ch2]] += 1
P = N.float() / N.sum(dim=1, keepdim=True)
```

```python
# calculate loss
def negative_log_likelihood(P, data):
    val = 0
    count = 0
    for ch1, ch2 in zip(data, data[1:]):
        val -= math.log(P[ctoi[ch1], ctoi[ch2]])
        count += 1
    val /= count
    return val

negative_log_likelihood(P=P, data=train), negative_log_likelihood(P=P, data=val)
```

```python
# prediction
output = ['\n']
max_output_length = 1000
while len(output) < max_output_length:
    i = ctoi[output[-1]]
    j = torch.multinomial(input=P[i], num_samples=1, replacement=True).item()
    output.append(itoc[j])
```

```python
print(''.join(output))
```
