```python
import torch
import torch.nn as nn
```

```python
# input & output initialization
x = torch.tensor([1, 2, 3, 4, 5], dtype=torch.float32)
print(f"x: {x}")
y = torch.tensor([1], dtype=torch.float32) # 1 if increasing, 0 if decreasing
print(f"y: {y}")
```

```python
# model declaration
class Perceptron(nn.Module):
    def __init__(self, input_size, output_size):
        super().__init__()
        self.head = nn.Linear(in_features=input_size, out_features=output_size)

    def forward(self, input):
        return self.head(input)
```

```python
# model initialization
model = Perceptron(input_size=x.shape[-1], output_size=y.shape[-1])
print(f"model.head.weight: {model.head.weight}")
print(f"model.head.bias: {model.head.bias}")
```

```python
# forward pass
with torch.inference_mode():
    z = model(x)
# activation function (value > 0 is 1, else 0)
y_pred = (z > 0).float()
print(f"y_pred: {y_pred.view(-1)}")
print(f"y:      {y.view(-1)}")
```
