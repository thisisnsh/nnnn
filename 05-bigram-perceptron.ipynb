```python
import torch
import torch.nn as nn
```

```python
# input & output initialization
with open("inputs-shakespeare.txt", "r") as f:
    data = f.read()
chars = sorted(set(data))
vocab_size = len(chars)
ctoi = {c: i for i, c in enumerate(chars)}
itoc = {i: c for i, c in enumerate(chars)}
```

```python
train = data[:int(len(data) * 0.9)]
val = data[int(len(data) * 0.9):]

x_train = torch.tensor([ctoi[ch] for ch in train[:-1]], dtype=torch.long)
y_train = torch.tensor([ctoi[ch] for ch in train[1:]], dtype=torch.long)
x_val = torch.tensor([ctoi[ch] for ch in val[:-1]], dtype=torch.long)
y_val = torch.tensor([ctoi[ch] for ch in val[1:]], dtype=torch.long)
```

```python
# input encoding
x_train_one_hot = nn.functional.one_hot(x_train, num_classes=vocab_size).float()
x_val_one_hot = nn.functional.one_hot(x_val, num_classes=vocab_size).float()
```

```python
# model declaration
class Perceptron(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.layer = nn.Linear(in_features=vocab_size, out_features=vocab_size, bias=False)

    def forward(self, x):
        return self.layer(x)
```

```python
# model initialization
model = Perceptron(vocab_size)
# optimizer initialization
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-1)
```

```python
# training
steps = 200
for step in range(steps):
    optimizer.zero_grad()
    
    logits_train = model(x_train_one_hot)
    loss_train = nn.functional.cross_entropy(logits_train, y_train)
    
    with torch.no_grad():
        logits_val = model(x_val_one_hot)
        loss_val = nn.functional.cross_entropy(logits_val, y_val)

    if step % 50 == 0:
        print(f"Step {step}: train loss = {loss_train.item()}, val loss = {loss_val.item()}")

    loss_train.backward()
    optimizer.step()
```

```python
# prediction
with torch.inference_mode():
    output = ['\n']
    max_output_length = 1000
    while len(output) < max_output_length:
        i = ctoi[output[-1]]
        
        one_hot = nn.functional.one_hot(torch.tensor(i), num_classes=vocab_size).float()
        logits = model(one_hot)
        P = torch.softmax(logits, dim=0)
        
        j = torch.multinomial(P, num_samples=1, replacement=True).item()

        output.append(itoc[j])
```

```python
print(''.join(output))
```
