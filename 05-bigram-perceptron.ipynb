```python
import torch
import torch.nn as nn
```

```python
# input & output initialization
with open(file="inputs-shakespeare.txt", mode="r", encoding="utf-8") as f:
    data = f.read()

chars = sorted(set(data))
vocab_size = len(chars)
ctoi = {char: i for i, char in enumerate(iterable=chars)}
itoc = dict(enumerate(iterable=chars))

split = int(len(data) * 0.9)
train, val = data[:split], data[split:]

x_train = torch.tensor([ctoi[ch] for ch in train[:-1]], dtype=torch.long)
y_train = torch.tensor([ctoi[ch] for ch in train[1:]], dtype=torch.long)
x_val = torch.tensor([ctoi[ch] for ch in val[:-1]], dtype=torch.long)
y_val = torch.tensor([ctoi[ch] for ch in val[1:]], dtype=torch.long)
```

```python
# input encoding
x_train_one_hot = nn.functional.one_hot(input=x_train, num_classes=vocab_size).float()
x_val_one_hot = nn.functional.one_hot(input=x_val, num_classes=vocab_size).float()
```

```python
# model declaration
class Perceptron(nn.Module):
    def __init__(self):
        super().__init__()
        self.head = nn.Linear(in_features=vocab_size, out_features=vocab_size, bias=False)

    def forward(self, input):
        return self.head(input=input)
```

```python
# model initialization
model = Perceptron()
# optimizer initialization
optimizer = torch.optim.AdamW(params=model.parameters(), lr=1e-1)
```

```python
# training
steps = 200
for step in range(steps):
    optimizer.zero_grad()
    
    train_logits = model(input=x_train_one_hot)
    train_loss = nn.functional.cross_entropy(input=train_logits, target=y_train)
    
    with torch.no_grad():
        val_logits = model(input=x_val_one_hot)
        val_loss = nn.functional.cross_entropy(input=val_logits, target=y_val)

    if step % 50 == 0:
        print(f"Step {step}: train loss = {train_loss.item()}, val loss = {val_loss.item()}")

    train_loss.backward()
    optimizer.step()
```

```python
# prediction
with torch.inference_mode():
    output = ['\n']
    max_output_length = 1000
    while len(output) < max_output_length:
        i = ctoi[output[-1]]
        
        one_hot = nn.functional.one_hot(input=torch.tensor(i), num_classes=vocab_size).float()
        logits = model(input=one_hot)
        P = torch.softmax(input=logits, dim=0)
        
        j = torch.multinomial(input=P, num_samples=1, replacement=True).item()

        output.append(itoc[j])
```

```python
print(''.join(output))
```
